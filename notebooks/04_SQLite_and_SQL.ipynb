```python
import pandas as pd
import mysql.connector
import os
from dotenv import load_dotenv

load_dotenv()

listings = pd.read_csv("../data/clean/listings_clean.csv")

print("Dataset Loaded Successfully!")

print("\nRows:", listings.shape[0])
print("Columns:", listings.shape[1])
```

# Connect Mysql

- establish a connection between mysql and python

```python
connection= mysql.connector.connect(
    host="localhost",
    user="root",
    password=os.getenv("MYSQL_PASSWORD"),
    database="airbnb_market_analysis"
)

cursor=connection.cursor()
print("✅ Connected to MySQL Successfully!")
```

```python
#Verify the connection
cursor.execute("SELECT DATABASE();")

print(cursor.fetchone())
```

# Create an Analytics Dataset

The cleaned dataset contains several columns that are not required for business reporting.

```python
analytics_columns = [
    "id",
    "name",
    "host_id",
    "host_name",
    "host_is_superhost",
    "neighbourhood_cleansed",
    "property_type",
    "room_type",
    "accommodates",
    "bedrooms",
    "beds",
    "bathrooms_text",
    "price",
    "minimum_nights",
    "maximum_nights",
    "availability_365",
    "number_of_reviews",
    "review_scores_rating",
    "review_scores_accuracy",
    "review_scores_cleanliness",
    "review_scores_location",
    "review_scores_value"
]

analytics_df = listings[analytics_columns]

analytics_df.head()
```

```python
analytics_df.shape
```

```python
analytics_df.to_csv(
    "../data/clean/airbnb_analytics.csv",
    index=False
)

print("✅ Analytics dataset saved successfully!")
```

```python
import os
from dotenv import load_dotenv
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

load_dotenv("../.env")

db_url = URL.create(
    drivername="mysql+mysqlconnector",
    username="root",
    password=os.getenv("MYSQL_PASSWORD"),
    host="localhost",
    database="airbnb_market_analysis"
)

engine = create_engine(db_url)
```

```python
analytics_df.to_sql(
    name='listings',
    con=engine,
    if_exists='replace',
    index=False
)
print("✅ Data loaded into MySQL successfully!")
```
