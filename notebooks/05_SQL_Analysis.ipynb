# Q1. How many listings are available for each room type?
SELECT
    room_type,
    COUNT(*) AS total_listings
FROM listings
GROUP BY room_type
ORDER BY total_listings DESC;

# Q2. Which neighbourhoods have the highest number of listings?

SELECT
    neighbourhood_cleansed,
    COUNT(*) AS total_listings
FROM listings
GROUP BY neighbourhood_cleansed
ORDER BY total_listings DESC
LIMIT 10;

# Q3. Which neighbourhoods have the highest average nightly price?

SELECT
    room_type,
    ROUND(AVG(price),2) AS average_price
FROM listings
GROUP BY room_type
ORDER BY average_price DESC;

# Q4. Which listings are the most expensive?

SELECT
    name,
    neighbourhood_cleansed,
    room_type,
    price
FROM listings
ORDER BY price DESC
LIMIT 10;

# Q5. What is the average listing price based on the number of bedrooms?

SELECT
    accommodates,
    ROUND(AVG(price), 2) AS average_price,
    COUNT(*) AS total_listings
FROM listings
GROUP BY accommodates
ORDER BY accommodates;

# Q6. Which hosts own the most Airbnb listings?
 
SELECT
    host_id,
    host_name,
    COUNT(*) AS total_listings
FROM listings
GROUP BY host_id, host_name
ORDER BY total_listings DESC
LIMIT 10;

# Q7. Do Superhosts receive higher customer ratings?

SELECT
    host_is_superhost,
    ROUND(AVG(review_scores_rating), 2) AS average_rating,
    COUNT(*) AS total_listings
FROM listings
WHERE review_scores_rating IS NOT NULL
GROUP BY host_is_superhost;

# Q8. Which neighbourhoods have the highest customer ratings?

SELECT
    neighbourhood_cleansed,
    ROUND(AVG(review_scores_rating), 2) AS average_rating,
    COUNT(*) AS total_listings
FROM listings
WHERE review_scores_rating IS NOT NULL
GROUP BY neighbourhood_cleansed
HAVING COUNT(*) >= 20
ORDER BY average_rating DESC
LIMIT 10;

## Q11. Which room type generates the highest estimated revenue potential?

SELECT
    room_type,
    COUNT(*) AS total_listings,
    ROUND(AVG(price),2) AS average_price,
    ROUND(COUNT(*) * AVG(price),2) AS estimated_revenue
FROM listings
GROUP BY room_type
ORDER BY estimated_revenue DESC;


## Q12. What is the price range in each room type?

SELECT
    room_type,
    MIN(price) AS minimum_price,
    MAX(price) AS maximum_price,
    ROUND(AVG(price),2) AS average_price
FROM listings
GROUP BY room_type;

## Q13. How many unique hosts are on Airbnb?

SELECT
    COUNT(DISTINCT host_id) AS unique_hosts
FROM listings;

## Q14. Categorize listings into Budget, Standard, and Luxury.

SELECT
    CASE
        WHEN price < 150 THEN 'Budget'
        WHEN price BETWEEN 150 AND 300 THEN 'Standard'
        ELSE 'Luxury'
    END AS price_category,
    COUNT(*) AS total_listings,
    ROUND(AVG(price),2) AS average_price
 FROM listings
 GROUP BY price_category
 ORDER BY average_price;

## Q15. Which neighbourhood has the highest average price among neighbourhoods with at least 100 listings?

SELECT
    neighbourhood_cleansed,
    COUNT(*) AS total_listings,
    ROUND(AVG(price),2) AS average_price
 FROM listings
 GROUP BY neighbourhood_cleansed
 HAVING COUNT(*) >= 100
 ORDER BY average_price DESC;

## Q16. Which listings are ranked among the most expensive within each neighbourhood?
SELECT
    neighbourhood_cleansed,
    name,
    room_type,
    price,
    RANK() OVER (
        PARTITION BY neighbourhood_cleansed
        ORDER BY price DESC
    ) AS price_rank
FROM listings
WHERE price IS NOT NULL;

## Q17. What percentage of listings belong to each room type?
SELECT
    room_type,
    COUNT(*) AS total_listings,
    ROUND(
        COUNT(*) * 100.0 /
        (SELECT COUNT(*) FROM listings),
        2
    ) AS percentage_of_listings
FROM listings
GROUP BY room_type
ORDER BY percentage_of_listings DESC;

## Q18. Which neighbourhoods have both high prices and strong customer ratings?
SELECT
    neighbourhood_cleansed,
    COUNT(*) AS total_listings,
    ROUND(AVG(price), 2) AS average_price,
    ROUND(AVG(review_scores_rating), 2) AS average_rating
FROM listings
WHERE price IS NOT NULL
  AND review_scores_rating IS NOT NULL
GROUP BY neighbourhood_cleansed
HAVING COUNT(*) >= 100
ORDER BY average_price DESC;
